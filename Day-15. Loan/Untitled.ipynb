```python
import pandas as pd
```

```python
loan = pd.read_csv('loan.csv', sep=',')
loan
```

```python
loan.columns
```

```python
loan['Dependents'] = loan['Dependents'].replace('3+', '4')
```

```python
loan['Dependents'] = pd.to_numeric(loan['Dependents'], errors='coerce')
```

```python
loan['Loan_Status'].dtype
```

```python
loan['Loan_Status'] = (
    loan['Loan_Status']
    .astype(str)
    .str.strip()
    .str.lower()
    .map({'y':1, 'yes':1, 'n':0, 'no':0})
)
```

```python
loan['LoanAmount'] = loan['LoanAmount'].fillna(0)
```

```python
loan
```

```python
loan['Gender'].hasnans
```

```python
loan['Gender'] = loan['Gender'].fillna('Female')
```

```python
loan['Married'] = loan['Married'].fillna('No')
```

```python
loan['Dependents'] = loan['Dependents'].fillna(0)
```

```python
loan['Loan_Amount_Term'] = loan['Loan_Amount_Term'].fillna(0)
```

```python
loan['Credit_History'] = loan['Credit_History'].fillna(1.0)
```

```python
loan['Self_Employed'] = loan['Self_Employed'].replace(' ', 'No')
```

```python
loan['Education'] = loan['Education'].replace(' ', 'not Graduate')
```

```python
loan['ApplicantIncome'] = loan['ApplicantIncome'].fillna(0)
```

```python
loan['CoapplicantIncome'] = loan['CoapplicantIncome'].fillna(0)
```

```python
loan['LoanAmount'] = loan['LoanAmount'].fillna(0)
```

```python
loan['Property_Area'] = loan['Property_Area'].fillna('Rural')
```

```python
loan['Gender'].hasnans
```

```python
loan.to_csv('loan_def.csv')
```
